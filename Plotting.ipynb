```python
import numpy as np

import os

import matplotlib.pyplot as plt
import matplotlib as mpl
from matplotlib.colors import SymLogNorm
from baryonification.baryonification.params import par
```

```python
par = par()

if par.code.multicomp:
    comp = 'multicomp' #'multicomp' or 'singlecomp'
else:
    comp = 'singlecomp'

size = par.code.size #'big' or 'small'

etacga1k = np.load('results_{}_{}/k_etacga=0.4.npy'.format(comp, size))
etacga1r = np.load('results_{}_{}/ratio_etacga=0.4.npy'.format(comp, size))

etacga2k = np.load('results_{}_{}/k_etacga=0.5.npy'.format(comp, size))
etacga2r = np.load('results_{}_{}/ratio_etacga=0.5.npy'.format(comp, size))

etacga3k = np.load('results_{}_{}/k_etacga=0.6.npy'.format(comp, size))
etacga3r = np.load('results_{}_{}/ratio_etacga=0.6.npy'.format(comp, size))

etacga4k = np.load('results_{}_{}/k_etacga=0.7.npy'.format(comp, size))
etacga4r = np.load('results_{}_{}/ratio_etacga=0.7.npy'.format(comp, size))

etacga5k = np.load('results_{}_{}/k_etacga=0.8.npy'.format(comp, size))
etacga5r = np.load('results_{}_{}/ratio_etacga=0.8.npy'.format(comp, size))
```

```python
# print the shot noise subtracted P(k)
plt.axhline(y = 1.0, color = 'black', linestyle = '-') 
plt.loglog(etacga1k.real, etacga1r.real, color = 'maroon',label = r'$\eta_{cga}=0.4$');
plt.loglog(etacga2k.real, etacga2r.real, color = 'red',label = r'$\eta_{cga}=0.5$');
plt.loglog(etacga3k.real, etacga3r.real, color = 'magenta',label = r'$\eta_{cga}=0.6$');
plt.loglog(etacga4k.real, etacga4r.real, color = 'tab:pink',label = r'$\eta_{cga}=0.7$');
plt.loglog(etacga5k.real, etacga5r.real, color = 'tab:purple',label = r'$\eta_{cga}=0.8$');

# format the axes
plt.xlabel(r"$k$ [$h \ \mathrm{Mpc}^{-1}$]");
# plt.ylabel(r"$k^{3}P(k)$");
plt.ylabel(r"$\frac{P_{displaced}(k)}{P_{non-displaced}(k)}$");
plt.xlim([np.min(etacga1k.real),kmax])
plt.legend()
plt.savefig(os.path.join('results_{}_{}'.format(comp, size), 'varying_etacga.pdf'), bbox_inches='tight')
```
